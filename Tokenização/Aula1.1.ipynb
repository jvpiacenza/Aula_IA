```python
vocabulario = ["bom","dia","ruim","muito"]

promt = "muito ruim"

tokens = []
for palavra in promt.split():
    if palavra in vocabulario:
        posicao = vocabulario.index(palavra)
        tokens.append(posicao)

print(tokens)
```

```python
#definindo pesos para cada index em ordem do vocabulario
pesos = [5,10,-10,2]

foco = "muito"
contexto = "ruim"

score = pesos[vocabulario.index(foco)]*pesos[vocabulario.index(contexto)]

print(f"Força da conexão é {score}")

```

```python
forca_total = 0

for t in tokens:
    forca_total += pesos[t]

    #print(forca_total)
if forca_total > 0:
    print("Positivo")
else:
    print("Negativo")
```
