```python
# Célula 2 - Análise de Sentimentos com SpaCy + TextBlob
import spacy
from textblob import TextBlob

# Carrega o modelo pré-treinado em inglês
nlp = spacy.load("en_core_web_sm")

# Frase de exemplo em inglês (TextBlob funciona melhor com inglês)
frase = "Yesterday I was happy, but Guilherme was sad and missed Minecraft and cried because he couldn't play"

# Processamento com SpaCy
doc = nlp(frase)

print("=== Tokens processados pelo SpaCy ===")
for token in doc:
    print(f"  Token: {token.text:15} | Lema: {token.lemma_:15} | POS: {token.pos_}")

# Análise de sentimento com TextBlob
blob = TextBlob(frase)
polaridade = blob.sentiment.polarity        # -1.0 (negativo) a +1.0 (positivo)
subjetividade = blob.sentiment.subjectivity # 0.0 (objetivo) a 1.0 (subjetivo)

print("\n=== Análise de Sentimento (TextBlob) ===")
print(f"Polaridade:    {polaridade:.2f}")
print(f"Subjetividade: {subjetividade:.2f}")

# Classificação: Positivo, Negativo ou Neutro
if polaridade > 0:
    resultado = "POSITIVO 😊"
elif polaridade < 0:
    resultado = "NEGATIVO 😢"
else:
    resultado = "NEUTRO 😐"

print(f"\nSentimento da frase: {resultado}")
```
