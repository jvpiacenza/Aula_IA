```python
#Vocabulario

lista_palavras_feliz = ["alegre","felicidade","contente","legar","massa"]
lista_palavras_triste = ["tristeza","chorar","magoado","saudade"]

talvez_feliz = 0
talvez_triste = 0
```

```python
#Tratamento da entrada da frase

frase = "Ontem eu estava alegre, pois o Guilherme estava com saudade do Minecraft e magoado por não conseguir jogar sem chorar"

frase_tratada = frase.lower().replace(".","").replace(",","")

palavras = frase_tratada.split()

#print(palavras)

for token in palavras:
    if token in lista_palavras_feliz:
        talvez_feliz += 1
    elif token in lista_palavras_triste:
        talvez_triste +=1

if talvez_triste > talvez_feliz:
    print("Essa frase é mais triste")
elif talvez_feliz > talvez_triste:
    print("Essa frase é mais feliz")

print((f"O resltado triste: {talvez_triste}"))
print((f"O resultado feliz: {talvez_feliz}"))
```
