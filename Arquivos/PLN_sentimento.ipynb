```python
# Célula 2 - Análise com tradução automática
import spacy
from textblob import TextBlob
from googletrans import Translator

# Carrega o modelo
nlp = spacy.load("en_core_web_sm")
translator = Translator()

# ✅ Frase em PORTUGUÊS
frase_pt = "O filme foi ótimo mas o final me deixou triste e decepcionado."

# Traduz para inglês automaticamente
traducao = translator.translate(frase_pt, src='pt', dest='en')
frase_en = traducao.text

print("=== Tradução ===")
print(f"Português: {frase_pt}")
print(f"Inglês:    {frase_en}")

# Processamento com SpaCy
doc = nlp(frase_en)

print("\n=== Tokens processados pelo SpaCy ===")
for token in doc:
    print(f"  Token: {token.text:15} | Lema: {token.lemma_:15} | POS: {token.pos_}")

# Análise de sentimento com TextBlob
blob = TextBlob(frase_en)
polaridade = blob.sentiment.polarity
subjetividade = blob.sentiment.subjectivity

print("\n=== Análise de Sentimento (TextBlob) ===")
print(f"Polaridade:    {polaridade:.2f}")
print(f"Subjetividade: {subjetividade:.2f}")

# Classificação
if polaridade > 0:
    resultado = "POSITIVO 😊"
elif polaridade < 0:
    resultado = "NEGATIVO 😢"
else:
    resultado = "NEUTRO 😐"

print(f"\nSentimento da frase: {resultado}")
```
